## 1. The North American Power Grid

The North American power grid is one of the largest and most complex engineered systems ever built. It spans three countries, stretches over millions of square miles, and delivers electricity to around **160 million customers** through more than **500,000 miles of high-voltage transmission lines**. The system operates at a nominal frequency of **60 Hz** — every generator, every motor, every device on the grid cycles in lockstep at that rate.

But here is the first thing to understand: **there is no single North American grid.** What most people call "the grid" is actually a collection of separate synchronous systems stitched loosely together.

### 1.1 How It Got This Way — A Brief History

In the early 20th century, electricity was a local business. Each city had its own utility, its own generators, its own wires. Starting in the 1920s, utilities began forming joint agreements to share peak-load coverage and backup power — the beginning of interconnection.

A key turning point came in **1965**, when a single relay failure in Ontario triggered a cascade that blacked out the entire northeastern US and parts of Canada — 30 million people lost power for up to 13 hours. The industry's response was to form **NERC** (originally the National Electric Reliability Council) in 1968, the first coordinated effort to set reliability standards across utilities.

For a brief period in the late 1960s, the Eastern and Western grids were actually connected via AC ties. But AC connections across such distances are prone to oscillations and instability, so those links were replaced with DC ties in **1975**, effectively splitting the continent into separate synchronous islands — the structure that still exists today.

```{note}
**AC tie vs. DC tie**

An **AC tie** is a transmission line that carries alternating current between two regions. Because AC systems must operate at exactly the same frequency and in phase, an AC tie *forces* the two regions into the same synchronous machine — they rise and fall together. This is fine for neighboring systems but causes instability problems over long distances, where slight frequency differences can trigger uncontrollable oscillations.

A **DC tie** (also called an HVDC link — high-voltage direct current) converts AC to DC at one end and back to AC at the other. Because DC carries no frequency information, it can connect two systems running at different frequencies (or different phase angles) without forcing them to synchronize. Power transfer is set by the operator's control system, not by physics. This makes DC ties a controlled, firewall-like interconnection: a fault or frequency event on one side does not automatically propagate to the other.
```

The **Energy Policy Act of 1992** was another turning point. Before it, a typical utility was a vertical monopoly — it owned the power plants, the transmission wires, *and* the local distribution lines, with no obligation to let anyone else use its infrastructure. The 1992 Act forced transmission owners to provide **open, nondiscriminatory access** to their networks. This cracked the monopoly model open and set the stage for competitive wholesale electricity markets — and for the ISOs and RTOs that manage those markets today.

### 1.2 The Five Interconnections

Today the grid is divided into **five wide-area synchronous interconnections**. Within each one, all generators rotate in sync at 60 Hz — if you measure the frequency anywhere inside the Eastern Interconnection right now, you get the same number as anywhere else inside it. Machines in *different* interconnections are not synchronized with each other.

The two majors are the **Eastern** and **Western** Interconnections, which together cover most of the continent. The Eastern alone has about **700 GW** of generating capacity (US portion); the Western has around **250 GW**. The three smaller ones are:

- **Texas Interconnection (ERCOT)** — covers most of Texas; tied to the Eastern grid by two DC links and to Mexico by a DC tie and a variable-frequency transformer (VFT)
- **Quebec Interconnection** — covers the entire Province of Quebec, operated as a single entity by Hydro-Québec; connected to the Eastern grid by four DC ties and one VFT
- **Alaska Interconnection** — actually two small, physically isolated grids with no connection to the rest of North America

The Eastern and Western grids are connected by **seven HVDC (high-voltage direct current) ties**, each around 200 MW, for a combined transfer capacity of about **1.32 GW**. That sounds significant until you realize it is less than 0.2% of the Eastern grid's capacity. For all practical purposes, the two major grids operate independently.

Why DC instead of AC for these cross-interconnection links? Because DC does not require synchronization — you can link two AC systems running at slightly different frequencies through a DC line without forcing them to lock together. This gives operators a controlled, stable way to move power between otherwise isolated systems.

### 1.3 Who Runs the Grid — Regulators and Operators

Grid governance in North America has two distinct layers: **reliability regulators** who set the rules, and **system operators** who run the grid day-to-day.

**FERC** (Federal Energy Regulatory Commission) is the US federal agency that regulates interstate electricity transmission and wholesale power markets. It does not operate the grid directly, but it establishes the framework under which ISOs and RTOs are authorized and wholesale markets are structured.

**NERC** (North American Electric Reliability Corporation), formed in 2006 as the successor to the original 1968 council, develops and enforces mandatory reliability standards for the bulk power system across the US, Canada, and parts of Mexico. Think of FERC as the *market* regulator and NERC as the *reliability* regulator — they cover complementary aspects of grid governance. Below NERC sit eight Regional Reliability Entities that handle compliance within their respective footprints.

For day-to-day operations, the grid is managed by **ISOs** (Independent System Operators) and **RTOs** (Regional Transmission Organizations). Both coordinate dispatch, manage transmission access, and run wholesale electricity markets. RTOs typically cover multi-state areas while ISOs tend to be single-state, though the distinction is largely historical. The major ones:

- **PJM** — 13 states + DC, from Illinois to New Jersey; the largest RTO in the world by peak load
- **MISO** — Midwest and parts of the South; covers about 75,000 miles of transmission
- **CAISO** — California
- **NYISO** — New York
- **ISO-NE** — the six New England states
- **ERCOT** — most of Texas; because it does not cross state lines, it is largely outside FERC jurisdiction
- **SPP** — Oklahoma, Kansas, and surrounding states

Large parts of the country — much of the Southeast and Pacific Northwest — are not covered by any ISO or RTO. There, **vertically integrated utilities** still own generation, transmission, and distribution together and operate their own systems without a centralized market.

### 1.4 Who Owns the Infrastructure

The physical infrastructure — power plants, transmission lines, substations — is owned by hundreds of utilities: investor-owned companies, municipal utilities, and rural electric cooperatives. The ten largest investor-owned utilities by market cap in 2025 are NextEra Energy (\$169B), Constellation Energy (\$114B), Southern Company (\$96B), Duke Energy (\$91B), American Electric Power (\$62B), Dominion Energy (\$51B), Exelon (\$44B), Xcel Energy (\$44B), Entergy (\$42B), and Consolidated Edison (\$36B). Duke has the largest annual revenue; NextEra is the world's largest producer of wind and solar energy.

## 2. Grid Structure

The traditional power grid had a simple, one-directional architecture: large power plants generated electricity, high-voltage transmission lines carried it over long distances, and distribution networks delivered it to homes and businesses. Power always flowed in one direction — from generator to customer.

That model is changing. Rooftop solar, battery storage, and electric vehicles allow customers to generate and push power *back* into the grid. This **two-way power flow** is one of the defining challenges — and opportunities — for modern grid optimization. Before we get into the math, it helps to understand what each layer of the grid actually does.

### 2.1 Generation

Generation is where electrical energy is produced. Traditionally this meant large, centralized **synchronous generators** — steam turbines burning coal or natural gas, or hydro turbines driven by falling water. These machines spin heavy rotors at a fixed speed (3600 rpm for 60 Hz in the US), and that spinning mass gives the grid a crucial property: **inertia**. When a sudden imbalance occurs between generation and load, the kinetic energy stored in all those spinning rotors buys the system time to respond — frequency droops slowly rather than collapsing instantly.

The generation mix has shifted substantially over the past two decades. Natural-gas-fired **combined-cycle** units (heat rates under 5,800 BTU/kWh) have largely replaced coal and oil plants, offering better efficiency, lower emissions, and faster ramp rates. But the more consequential shift is the rise of **wind and solar**. Both produce zero-emission electricity at costs that have plummeted, but neither has a spinning rotor — they connect to the grid through **inverters**, which means they contribute little or no inertia. As inverter-based generation displaces conventional generators, the grid's frequency response becomes less forgiving.

At a smaller scale, **distributed energy resources (DERs)** — rooftop solar panels, behind-the-meter batteries, small combined heat-and-power units — sit on the distribution system rather than the transmission system. Their aggregate capacity is growing fast and complicates the traditional picture: generation is no longer something that only happens at large plants on the transmission grid.

### 2.2 Transmission

The transmission system is the high-voltage backbone of the grid. Its job is to move bulk power efficiently over long distances — from remote generation sites (mountain wind farms, desert solar plants, large hydro facilities) to load centers (cities, industrial regions).

Transmission lines operate at high voltages — typically **115 kV to 765 kV** in North America — because higher voltage means lower current for the same power, which means lower resistive losses. A transformer at the generating station steps voltage up for long-distance travel; another transformer at the receiving substation steps it back down for local distribution.

The transmission system is a **meshed network**: most major substations are connected to multiple lines, so power can take many parallel paths from source to sink. This redundancy is essential for reliability — if one line trips, power reroutes through others. But it also means power flows are governed by the physics of the network (Kirchhoff's laws), not by the preferences of operators. You cannot simply tell power to flow along a specific path; it distributes itself according to the impedances of all available paths. This is precisely what makes transmission planning and operation a hard optimization problem.

Two technologies are increasingly important on the transmission system:

**HVDC (High-Voltage Direct Current)** links are used where AC transmission is impractical — very long distances, submarine cables, or connections between asynchronous systems (like the Eastern and Western Interconnections). HVDC allows precise, controllable power transfer and does not suffer from AC stability issues like oscillations. Modern **Voltage Source Converter (VSC) HVDC** systems can also provide reactive power support, making them even more flexible.

**FACTS (Flexible AC Transmission Systems)** are power-electronics devices that dynamically control voltage and power flow on AC lines. Shunt devices like the **SVC** (Static Var Compensator) and **STATCOM** (Static Synchronous Compensator) inject or absorb reactive power to regulate voltage and improve stability. Series devices like the **TCSC** (Thyristor-Controlled Series Compensator) change effective line impedance to redirect power flows and damp oscillations. FACTS devices give operators real-time controllability over a network that would otherwise be governed entirely by passive physics.

### 2.3 Distribution

The distribution system is the last mile — the medium- and low-voltage network (typically **4 kV to 35 kV** for medium voltage, **120/240 V** for residential service) that connects substations to individual customers. It is radial by design: power flows outward from a substation along feeders, branching into progressively smaller lines until it reaches homes and businesses.

The distribution system was originally built to be **passive**: it delivered power in one direction and required little real-time control. A substation transformer stepped voltage down, breakers and fuses protected against faults, and that was largely it. Utilities planned distribution networks for peak demand and rarely needed to think about what happened in between.

That simplicity is gone. The proliferation of rooftop solar, home batteries, EV chargers, and smart appliances means distribution feeders now experience **bidirectional power flows** and highly variable net loads. A feeder that once saw a smooth daily demand curve now sees sharp ramps driven by cloud cover and EV charging patterns. Voltage can rise unexpectedly in the middle of the day when solar generation exceeds local demand. Equipment originally designed for passive, one-way operation must now handle these conditions.

Modern distribution systems respond with **distribution automation**: SCADA systems that monitor the network in real time, automated switches that can isolate faults and restore service within seconds (FLISR — Fault Location, Isolation, and Service Restoration), and smart inverters that can actively regulate voltage by absorbing or injecting reactive power.

**Microgrids** represent an emerging layer below the distribution system. A microgrid is a small, self-contained network — a university campus, a military base, a neighborhood — that can operate either connected to the main grid or as an island during outages. Microgrids run their own local optimization: balancing local generation (solar, diesel, storage) against local demand, deciding when to buy from or sell to the grid, and maintaining power quality when islanded. As microgrids multiply, the distribution system increasingly resembles a network of interacting sub-systems rather than a passive delivery infrastructure.

### 2.4 The Changing Boundary

For most of the grid's history, transmission and distribution were treated as separate worlds. Transmission planners handed load forecasts to distribution planners and that was the extent of the interaction. Distribution operators assumed an "infinite bus" at the substation — a fixed voltage source that would always supply whatever they needed.

That assumption is breaking down. As DERs aggregate to meaningful scale, the distinction between "generation" and "load" blurs — a feeder with heavy solar becomes a net exporter during midday, a net importer at night. Transmission operators can no longer ignore what is happening on the distribution system, and distribution operators can no longer ignore transmission-level signals. **Integrated T&D planning and operation** is one of the major open research problems in power systems today — and it is fundamentally an optimization problem.

## 3. Key Components

To build optimization models for the grid, we need to move from the big-picture view of Section 2 down to the individual components — their physical behavior, their parameters, and how we represent them mathematically. This section introduces the five building blocks that appear in almost every power system model.

### 3.1 Generators

A generator converts mechanical or chemical energy into electrical energy. For optimization purposes, we do not model the turbine physics in detail — we treat each generator as a **controllable source** characterized by a handful of parameters.

The most important are its **active power output** $P_g$ (measured in MW) and **reactive power output** $Q_g$ (measured in MVAr). These are the two decision variables the system operator can adjust. They are not free to take any value; each generator has:

- A **maximum and minimum active power output**: $P_g^{\min} \le P_g \le P_g^{\max}$. The minimum is often nonzero for thermal units — a coal plant cannot run at 1% capacity.
- A **reactive power capability**: $Q_g^{\min} \le Q_g \le Q_g^{\max}$. The limits depend on the current active output and the machine's design, tracing out a region called the **capability curve**.
- A **ramp rate**: how fast the output can change per unit time (MW/min). Critical for unit commitment and real-time dispatch.
- A **cost function**: for thermal units, the fuel cost as a function of output is typically approximated as a quadratic:
$$C_g(P_g) = a_g P_g^2 + b_g P_g + c_g \quad [\$/\text{h}]$$
Minimizing the total generation cost subject to the network constraints is, in its simplest form, the **Optimal Power Flow (OPF)** problem — the central topic of this notebook series.

Wind and solar generators are modeled differently: their active power output is bounded above by the available resource (which is uncertain and time-varying) rather than a fixed $P_g^{\max}$, and they typically have near-zero marginal cost.

### 3.2 Transmission Lines

A transmission line carries power between two buses. Its electrical behavior is determined by three distributed physical properties: **resistance** $r$ (causes real power loss), **inductance** $x$ (causes reactive power interaction), and **capacitance** $b$ (causes reactive power generation at light loads). For a line of given length, these combine into lumped parameters $R$, $X$, and $B$.

In power flow models, a transmission line is represented by the **$\pi$-model**: a series branch with impedance $z = R + jX$ connecting the two endpoints, and two shunt capacitors $B/2$ hanging to ground at each end.

```
Bus i ──[R + jX]── Bus k
  |                    |
 jB/2               jB/2
  |                    |
 GND                 GND
```

The series admittance is $y = 1/z = g + jb$ where $g$ is conductance and $b$ is susceptance. For most high-voltage transmission lines, $R \ll X$, so resistance is sometimes neglected — this is the basis of the **DC power flow approximation** we will use extensively in later notebooks.

Each line also has a **thermal limit** — the maximum current it can carry before the conductor overheats. In power flow, this is typically expressed as a maximum apparent power (MVA) or active power (MW) flow, and appears as an inequality constraint in OPF.

### 3.3 Transformers

Transformers change voltage levels, allowing the same power to be transmitted at different voltages across different parts of the network. They are what make the multi-voltage-level structure of the grid (generation → transmission → distribution) possible.

In circuit terms, an ideal transformer with turns ratio $a : 1$ relates voltages and currents on its two sides by:
$$V_1 = a \, V_2, \qquad I_1 = I_2 / a$$

Real transformers also have leakage **reactance** $X_T$ (modeled as a series impedance), which limits fault currents and causes a small voltage drop under load. In power flow models, a transformer is treated as a transmission line with a series impedance and a **tap ratio** — a complex multiplier that sets the voltage transformation. Most transformers have a fixed tap ratio, but **on-load tap changers (OLTCs)** can adjust their ratio under load to regulate voltage, making them another controllable element in optimization problems.

### 3.4 Buses

A **bus** (short for busbar) is a node in the network — a point where transmission lines, generators, loads, and transformers connect. Physically, it is a section of conductor in a substation at which multiple circuits are joined.

Each bus $i$ has four associated quantities: voltage magnitude $|V_i|$, voltage angle $\theta_i$, net active power injection $P_i$, and net reactive power injection $Q_i$. In the power flow problem, two of these four are specified (known) and two are unknown. This classification defines three bus types:

| Bus type | Also called | Specified | Unknown |
|---|---|---|---|
| **Slack** | Reference | $|V|$, $\theta = 0$ | $P$, $Q$ |
| **PV** | Generator bus | $P$, $|V|$ | $Q$, $\theta$ |
| **PQ** | Load bus | $P$, $Q$ | $|V|$, $\theta$ |

The **slack bus** is a mathematical necessity: it serves as the angle reference ($\theta = 0$) and absorbs the small active and reactive power mismatches that arise from losses. In practice, the slack bus is typically the largest generator in the system, which has enough headroom to supply whatever is needed to balance the network.

**PV buses** are generator buses where the operator specifies active power output and holds voltage magnitude constant (via the generator's excitation system). **PQ buses** are load buses where real and reactive demand are fixed by the load and the bus voltage floats.

### 3.5 Loads

A **load** represents the aggregate electrical demand at a bus — the sum of all consumer devices connected at that point. In the simplest model, a load is treated as a **constant power demand**: it always consumes exactly $P_d$ MW of active power and $Q_d$ MVAr of reactive power regardless of voltage. This is the default assumption in most OPF formulations.

More detailed models allow load to depend on voltage. The **ZIP model** decomposes load into three components:

$$
P = P_0 \left[ Z_p \left(\frac{|V|}{|V_0|}\right)^2 + I_p \frac{|V|}{|V_0|} + P_p \right]
$$

where $Z_{p}$, $I_{p}$, $P_{p}$ are the fractions of constant-impedance (Z), constant-current (I), and constant-power (P) components, with $Z_{p} + I_{p} + P_{p} = 1$. At the nominal voltage $|V_0|$, the total is simply $P_0$. This model matters in distribution systems where voltage variations are larger.

**Demand response** treats load as a partially controllable resource: large industrial customers can voluntarily reduce consumption in exchange for compensation. In optimization, interruptible load appears as a variable with a cost (the value of lost load) rather than a fixed parameter. This blurs the boundary between generation-side and demand-side resources — a theme that runs through modern electricity market design.

## 4. Key Quantities

Before writing down a single power flow equation, we need to be precise about the quantities involved. Power systems use a specific set of electrical variables — and a specific convention for normalizing them — that can be confusing at first. This section builds that foundation.

### 4.1 Active and Reactive Power

In a DC circuit, power is simply $P = VI$. In an AC circuit, both voltage and current are sinusoids, and they may be **out of phase** with each other. This phase difference is what gives rise to the distinction between active and reactive power.

Consider a sinusoidal voltage and current at a bus:
$$
v(t) = V_{\max} \cos(\omega t), \qquad i(t) = I_{\max} \cos(\omega t - \varphi)
$$

The instantaneous power is $p(t) = v(t) \cdot i(t)$. Expanding this product and averaging over one full cycle gives two distinct components:

$$
P = \underbrace{V_{\text{rms}} I_{\text{rms}} \cos\varphi}_{\text{active power [W, MW]}} \qquad Q = \underbrace{V_{\text{rms}} I_{\text{rms}} \sin\varphi}_{\text{reactive power [VAr, MVAr]}}
$$

**Active power** $P$ is the time-averaged power that does real work — it runs motors, heats resistors, lights bulbs. **Reactive power** $Q$ is power that sloshes back and forth between source and load every half cycle, doing no net work but required to sustain the electric and magnetic fields inside inductive and capacitive devices (motors, transformers, cables).

These two are combined into the **complex power**:

$$S = P + jQ = V I^* \qquad [\text{VA, MVA}]$$

where $V$ and $I$ are phasors (complex numbers representing the amplitude and phase of the sinusoids) and $I^*$ is the complex conjugate of $I$. The magnitude $|S| = \sqrt{P^2 + Q^2}$ is called **apparent power**, and the ratio $\cos\varphi = P/|S|$ is the **power factor**.

**Why does reactive power matter for optimization?** Unlike active power, reactive power does not travel efficiently over long distances — it causes voltage drops along the way. Every bus in the network must have its reactive power supply and demand balanced locally; otherwise voltage will drift outside acceptable limits. Generators, capacitor banks, and FACTS devices all serve as sources of reactive power. Managing this balance — alongside the dispatch of active power — is a central constraint in AC-OPF.

A useful sign convention: **generation is positive** (injection into the network), **consumption is negative** (withdrawal). So a load with demand $P_d > 0$ appears as $-P_d$ in the nodal injection equations.

### 4.2 Voltage and Frequency

**Voltage** at each bus is represented as a complex phasor:
$$
V_i = |V_i| e^{j\theta_i} = |V_i|(\cos\theta_i + j\sin\theta_i)
$$

The **magnitude** $|V_i|$ must stay close to the nominal value (typically within ±5%) to ensure equipment operates correctly and reactive power can be delivered. Too low a voltage means motors stall and reactive demand spirals upward — a precursor to **voltage collapse**. Too high a voltage stresses insulation and may trip protective equipment.

The **angle** $\theta_i$ carries information about active power flows. For two buses $i$ and $k$ connected by a lossless line with reactance $X$, the active power flow is approximately:
$$
P_{ik} \approx \frac{|V_i||V_k|}{X} \sin(\theta_i - \theta_k)
$$

Power flows from the higher-angle bus to the lower-angle bus. Angles are always measured relative to the slack bus (which is set to $\theta = 0$ by convention), so only angle *differences* are physically meaningful.

**Frequency** is a system-wide quantity — within a synchronous interconnection, every machine operates at the same frequency at any given instant. Frequency reflects the instantaneous balance between total generation and total load:

- If generation **exceeds** load, rotors accelerate and frequency **rises** above 60 Hz.
- If load **exceeds** generation, rotors slow and frequency **falls** below 60 Hz.

The grid responds to frequency deviations in layers. The first response is **inertia**: spinning rotors immediately release or absorb kinetic energy, slowing the rate of frequency change. Within seconds, **governor droop control** adjusts generator outputs proportionally to the deviation. Over minutes, **Automatic Generation Control (AGC)** restores frequency to exactly 60 Hz by adjusting generator setpoints. Finally, **economic dispatch** re-optimizes the generation schedule. This notebook series focuses primarily on that last layer.

Normal operating range is roughly **59.95–60.05 Hz**. Protective relays begin shedding load automatically below about 59.3 Hz to prevent a total collapse.

### 4.3 Per-Unit System

Power systems span multiple voltage levels — a generator might produce at 20 kV, step up to 345 kV for transmission, then step down to 13.8 kV for distribution. Comparing impedances or currents across these levels in physical units (ohms, amps) is unwieldy. The **per-unit (p.u.) system** solves this by normalizing everything to dimensionless ratios.

The idea: choose a **base value** for each physical quantity, then express all values as a fraction of that base:
$$
x_{\text{p.u.}} = \frac{x_{\text{actual}}}{x_{\text{base}}}
$$

In practice, we freely choose two base quantities — a system-wide **apparent power base** $S_{\text{base}}$ and a **voltage base** $V_{\text{base}}$ at each voltage level — and derive the rest:

$$
I_{\text{base}} = \frac{S_{\text{base}}}{\sqrt{3}\, V_{\text{base}}}, \qquad Z_{\text{base}} = \frac{V_{\text{base}}^2}{S_{\text{base}}}
$$

A typical choice is $S_{\text{base}} = 100$ MVA system-wide, with $V_{\text{base}}$ set to the nominal voltage at each level (e.g., 345 kV on the transmission side, 20 kV on the generator side).

The key benefits of working in per-unit:

- **Transformer ratios disappear.** An ideal transformer with ratio $a:1$ has per-unit ratio $1:1$ if the voltage bases on each side are chosen to match the transformer ratio. This simplifies the network equations considerably.
- **All quantities are order-1.** A 345 kV bus at nominal voltage has $|V| = 1.0$ p.u. Acceptable voltages are 0.95–1.05 p.u. Line impedances are typically 0.001–0.1 p.u. This makes it easy to spot data errors.
- **Equations become cleaner.** The power flow equations look the same regardless of the physical voltage level.

Throughout this notebook series, all variables are in per-unit unless explicitly stated otherwise. When working with real data (MATPOWER cases, EGRET inputs), the conversion from physical to per-unit is handled automatically by the software.

## 5. Summary

This notebook provided the background needed to make sense of the optimization problems in later notebooks. The key takeaways:

**The grid is not one system.** North America operates five separate synchronous interconnections. The Eastern and Western are the two majors, connected only by a handful of HVDC ties. Within each interconnection, all machines run in sync at 60 Hz.

**The grid has three layers.** Generation produces power; the high-voltage transmission network moves it over long distances; the distribution system delivers it to customers. Traditionally these layers were one-way and largely independent. Rising DER penetration is breaking down both assumptions.

**Every component is a model.** Generators are controllable sources with cost functions and output limits. Transmission lines are π-models with series impedance and shunt susceptance. Transformers introduce tap ratios. Buses are nodes classified by which quantities are fixed and which are unknowns. Loads are (mostly) fixed power demands.

**Two kinds of power, two kinds of constraint.** Active power $P$ does real work and flows along angle gradients. Reactive power $Q$ sustains electromagnetic fields and governs voltage magnitudes. Both must be balanced at every bus, every instant — this coupling is what makes power flow nonlinear and AC-OPF a hard problem.

**Work in per-unit.** Normalize all quantities to a 100 MVA base. Transformer ratios vanish, all impedances become comparable, and the equations look the same regardless of voltage level.
